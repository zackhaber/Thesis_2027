# Environment demo
Generate a reproducible level, render it, and check one episode.

```python
from env import DifficultySpec, PursuitEvasionEnv, generate_level
from ui import render_ascii
scenario = generate_level(DifficultySpec(level=1, seed=7))
environment = PursuitEvasionEnv(scenario)
environment.reset()
print(render_ascii(environment))
```

```python
observation, reward, done, info = environment.step(['stay'] * len(scenario.pursuers), ['stay'] * len(scenario.evaders))
print(reward, done, info)
```
