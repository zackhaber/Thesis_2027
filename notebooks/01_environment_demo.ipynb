# Environment demo

This notebook introduces the pursuit-evasion environment. It generates a reproducible Level 1 scenario, resets the environment, and renders the initial state as an ASCII grid. It then performs one environment step with all agents choosing `stay` to demonstrate advancing the environment without changing positions, and displays the resulting rewards, termination status, and episode information.

Run the cells in order because the second code cell uses the `scenario` and `environment` objects created in the first code cell.

```python
from env import DifficultySpec, PursuitEvasionEnv, generate_level
from ui import render_ascii
scenario = generate_level(DifficultySpec(level=1, seed=7))
environment = PursuitEvasionEnv(scenario)
environment.reset()
print(render_ascii(environment))
```

```python
observation, reward, done, info = environment.step(['stay'] * len(scenario.pursuers), ['stay'] * len(scenario.evaders))
print(reward, done, info)
```
