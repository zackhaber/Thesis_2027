# Curriculum training

This notebook prepares scenarios at Levels 1, 2, and 3 to represent an increasing-difficulty curriculum. It then trains a QMIX-labelled agent for 30 episodes and reports the number of generated levels and the final reward.

The current code demonstrates the setup for curriculum learning; the generated levels are not yet passed into the training loop. As a result, the training run is not currently conditioned on those scenarios.

```python
from marl import build_agent
from env import DifficultySpec, generate_level
levels = [generate_level(DifficultySpec(level=level, seed=level)) for level in (1, 2, 3)]
run = build_agent('QMIX').train(episodes=30)
len(levels), run.rewards[-1]
```
