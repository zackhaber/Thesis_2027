# Curriculum training
Train or inspect runs in increasing difficulty order.

```python
from marl import build_agent
from env import DifficultySpec, generate_level
levels = [generate_level(DifficultySpec(level=level, seed=level)) for level in (1, 2, 3)]
run = build_agent('QMIX').train(episodes=30)
len(levels), run.rewards[-1]
```
